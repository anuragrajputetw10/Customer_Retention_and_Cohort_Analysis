1. Load Dataset

```python
import pandas as pd
import numpy as np

# Display all columns
pd.set_option("display.max_columns", None)

# Load dataset
df = pd.read_csv("D:\ANURAG\Customer Retention + Cohort Analysis project\kaggle dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# Display basic information
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
```

3. Missing Value Analysis

```python
# Check total missing values

missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
# Missing value percentage

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({"Missing Values": missing_values, "Missing Percentage": missing_percentage })

missing_summary
```

```python
# Rows with missing CustomerID

missing_customer = df[df["CustomerID"].isnull()].copy()

print("Rows with missing CustomerID:")
print(missing_customer.shape)

missing_customer.head()
```

```python
# Rows with missing Description

missing_description = df[df["Description"].isnull()].copy()

print("Rows with missing Description:")
print(missing_description.shape)

missing_description.head()
```

4. Cancelled Order Analysis

```python
# convert InvoiceNo to string.

df["InvoiceNo"] = df["InvoiceNo"].astype(str)
```

```python
# Identify cancelled orders

cancelled_orders = df[df["InvoiceNo"].str.startswith("C")].copy()

print("Total Cancelled Transaction Rows:")
print(cancelled_orders.shape[0])

cancelled_orders.head()
```

```python
# Number of unique cancelled invoices

print(
    "Unique Cancelled Orders:", cancelled_orders["InvoiceNo"].nunique())
```

```python
# Total cancellation value

# Because cancelled transactions generally have negative quantities:

cancelled_orders["Revenue"] = (cancelled_orders["Quantity"] * cancelled_orders["UnitPrice"])

print("Total Cancelled Value:",cancelled_orders["Revenue"].sum())
```

```python
print("Absolute Cancelled Value:",abs(cancelled_orders["Revenue"].sum()))
```

```python
# Preview cancellation data

cancelled_orders[
    [
        "InvoiceNo",
        "StockCode",
        "Description",
        "Quantity",
        "InvoiceDate",
        "UnitPrice",
        "CustomerID",
        "Country"
    ]
].head(10)
```

5. Invalid Transaction Analysis

```python
# Quantity less than or equal to zero

invalid_quantity = df[(~df["InvoiceNo"].str.startswith("C")) &(df["Quantity"] <= 0)].copy()

print("Invalid Quantity Rows:")
print(invalid_quantity.shape[0])

invalid_quantity.head()
```

```python
# UnitPrice less than or equal to zero

invalid_price = df[
    (~df["InvoiceNo"].str.startswith("C")) &
    (df["UnitPrice"] <= 0)
].copy()

print("Invalid Price Rows:")
print(invalid_price.shape[0])

invalid_price.head()
```

6. Duplicate Analysis

```python
# Check exact duplicate rows

duplicate_count = df.duplicated().sum()

print("Total Exact Duplicate Rows:")
print(duplicate_count)
```

```python
# Display duplicate rows

duplicates = df[
    df.duplicated()
].copy()

duplicates.head()
```

```python
# Percentage of duplicates

duplicate_percentage = ( df.duplicated().sum() / len(df)) * 100

print( f"Duplicate Percentage: {duplicate_percentage:.2f}%" )
```

7. Data Cleaning

```python
# Remove cancelled orders

sales_df = df[~df["InvoiceNo"].str.startswith("C")].copy()

print("After removing cancelled orders:")
print(sales_df.shape)
```

```python
sales_df
```

```python
# Separate missing CustomerID

missing_customer = sales_df[sales_df["CustomerID"].isnull()].copy()

print("Rows with missing CustomerID:",missing_customer.shape[0])
```

```python
# Remove them from the main dataset:

sales_df = sales_df.dropna(subset=["CustomerID"]).copy()

print("Shape after removing missing CustomerID:",sales_df.shape)
```

```python
# Separate invalid UnitPrice

invalid_price = sales_df[sales_df["UnitPrice"] <= 0].copy()

print("Invalid UnitPrice Rows:",invalid_price.shape[0])

sales_df = sales_df[sales_df["UnitPrice"] > 0].copy()

sales_df.shape
```

```python
# Remove duplicates

before_duplicates = sales_df.shape[0]

sales_df = sales_df.drop_duplicates().copy()

after_duplicates = sales_df.shape[0]

print("Duplicates Removed:",before_duplicates - after_duplicates)
```

```python
# Handle missing Description


print("Missing Description:",sales_df["Description"].isnull().sum())

# want to remove them:

# sales_df = sales_df.dropna(subset=["Description"])
```

```python
# Fix data types

sales_df["InvoiceDate"] = pd.to_datetime(sales_df["InvoiceDate"])

sales_df["CustomerID"] = (sales_df["CustomerID"].astype(int))

sales_df.dtypes
```

8. Feature Engineering

```python
# Revenue

sales_df["Revenue"] = (sales_df["Quantity"] * sales_df["UnitPrice"])
```

```python
# Invoice Date Features

# Year
sales_df["Year"] = (sales_df["InvoiceDate"].dt.year)

# Month Number
sales_df["Month"] = (sales_df["InvoiceDate"].dt.month)

# Month Name
sales_df["Month_Name"] = (sales_df["InvoiceDate"].dt.month_name())

# Year-Month
# This will be extremely useful for cohort analysis.

sales_df["Year_Month"] = (sales_df["InvoiceDate"].dt.to_period("M").astype(str))

# Day of Week
sales_df["Day_Name"] = (sales_df["InvoiceDate"].dt.day_name())

# Day of Month
sales_df["Day"] = (sales_df["InvoiceDate"].dt.day)

# Hour
# This can help analyze purchasing behavior by time.
sales_df["Hour"] = (sales_df["InvoiceDate"].dt.hour)
```

Final Data Quality Report Of SALES DATA

Now check whether the cleaned data has the issues we intended to remove.

```python
print("=" * 50)
print("FINAL DATA QUALITY REPORT")
print("=" * 50)

print("\nFinal Dataset Shape:")
print(sales_df.shape)

print("\nMissing Values:")
print(sales_df.isnull().sum())

print("\nExact Duplicate Rows:")
print(sales_df.duplicated().sum())

print("\nCancelled Orders:")
print(sales_df["InvoiceNo"].str.startswith("C").sum())

print("\nQuantity <= 0:")
print((sales_df["Quantity"] <= 0).sum())

print("\nUnitPrice <= 0:")
print((sales_df["UnitPrice"] <= 0).sum())

print("\nCustomerID Missing:")
print(sales_df["CustomerID"].isnull().sum())

print("\nTotal Revenue:")
print(f"£{sales_df['Revenue'].sum():,.2f}")

print("\nUnique Customers:")
print(sales_df["CustomerID"].nunique())

print("\nUnique Orders:")
print(sales_df["InvoiceNo"].nunique())

print("=" * 50)
print("DATA CLEANING COMPLETED")
print("=" * 50)
```
